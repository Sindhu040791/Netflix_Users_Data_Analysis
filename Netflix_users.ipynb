```python
import matplotlib.pyplot as plt
```

```python
result_table['Country'].value_counts().head(5).plot(kind='bar',color="purple")
plt.title('Number of Netflix Users by Country')
plt.xlabel('Country')
plt.ylabel('Number of Users')
plt.grid(False)
plt.show()
```

```python
plt.hist(result_table['Age'], bins=10,color="green", edgecolor='black')
plt.title('Age Distribution of Netflix Users')
plt.xlabel('Age')
plt.ylabel('Number of Users')
plt.show()
```

```python
result_table['Favorite_Genre'].value_counts().plot(kind='pie',autopct='%1.1f%%',startangle=90,colors=plt.cm.Paired.colors)
plt.title('Favorite Genres of Netflix Users')
plt.show()
```
